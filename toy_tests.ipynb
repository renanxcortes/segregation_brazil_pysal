```python
import geopandas as gpd
from shapely.geometry import Polygon
import pandas as pd

# 1. Create dummy polygon geometries
polygons = [
    Polygon([(0, 0), (1, 0), (1, 1), (0, 1)]),
    Polygon([(1, 0), (2, 0), (2, 1), (1, 1)]),
    Polygon([(0, 1), (1, 1), (1, 2), (0, 2)]),
    Polygon([(1, 1), (2, 1), (2, 2), (1, 2)])
]

# 2. Create data for the columns
data1 = {
    'district': ['A', 'B', 'C', 'D'],
    'total_pop': [1000, 1500, 800, 2000],
    'minority_pop': [20, 30, 15, 40] # All values are less than total_pop
}

# 3. Create the GeoDataFrame
gdf1 = gpd.GeoDataFrame(data1, geometry=polygons)
```

```python
# 2. Create data for the columns
data2 = {
    'district': ['A', 'B', 'C', 'D'],
    'total_pop': [1000, 1500, 800, 2000],
    'minority_pop': [20 * 2, 30 * 2, 15 * 2, 40 * 2] # All values are less than total_pop
}

# 3. Create the GeoDataFrame
gdf2 = gpd.GeoDataFrame(data2, geometry=polygons)
```
